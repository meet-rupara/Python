```python
# Type Casting 
# converting object of one data tyoe into other 

#Impicit Casting 

a=1234
b=567.8
c=a+b
print(c)
```

```python
a=True   
b=10
print(a+b)    # true is counted as one
```

```python
#Expicit casting 
#usinf in-built functions like int(), float(),str()

# using int() function 
a=int(123.45)
b=int(True)
c=int(False)
d=int("1110",2)  #Similarly for octal,hexadecimal  use 8,16 respectively 
e=int("10101")

print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
# float() method 

a=float(12)
b=float("1.1123E5") #scientific notation string
c=float(2.0)
print(a)
print(b)
print(c)

```

```python
# str() function 

a=str(100)
b=str(101.1)
c=str(1234e5)

print(a)
print(b)
print(c)

```

```python
a=str([3,5,8,6,])
b=str((3,8,5,9))
c=str({a:124,b:457,c:576})
print(a)
print(b)
print(c)
```
